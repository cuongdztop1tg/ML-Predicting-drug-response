```python
import pandas as pd

file_path = '../csa_data/raw_data/y_data/response.tsv'

print("Đang đọc file response...")
# Đọc file (nhớ low_memory=False để tránh cái lỗi Mixed types bạn gặp lúc nãy)
df = pd.read_csv(file_path, sep='\t', low_memory=False)

# 1. Tạo cột 'split_id' bằng chính chỉ số dòng (0, 1, 2, 3...)
# Các file split (như CCLE_split_0_train.txt) thường chứa các con số này
df['split_id'] = df.index

# 2. Lưu đè lại file
df.to_csv(file_path, sep='\t', index=False)

print("Đã thêm cột 'split_id' thành công!")
print(df[['split_id', 'source', 'improve_sample_id']].head())
```
